# matrix multiplication

```python
a=torch.tensor([
    [1,2],
    [3,4]
])
b=torch.tensor([
    [2,3],
    [1,4]
])
a,b
```

```python
a@b
```

```python
torch.matmul(a,b)
```

```python
a=torch.tensor([
    [1,2],
    [3,4],
    [6,6]
])
b=torch.tensor([
    [2,3],
    [1,4]
])
a.shape,b.shape
```

```python
c=a@b
c
```

```python
c.shape
```

```python
a=torch.tensor([
    [1,2,3],
    [3,4,9],

])
b=torch.tensor([
    [2,3],
    [1,4]
])
a.shape,b.shape
```

```python
a@b
```

```python
a.T.shape
```

```python
a.T@b
```

```python
# (3,2)@(2,2)-->(3,2)
torch.matmul(a.T,b)
```
